**The direct ask.** Our team was hired to find the optimal weights for a tangency portfolio built from 8 core stocks: AAPL, BA, T, MGM, AMZN, IBM, TSLA, and GOOG, with the assumption of having no transaction fees. The goal was to test whether utilizing a Markowitz-style approach lets a mutual fund manager beat the market. Using the full data sample (2012–2020, Task 6), the maximum-Sharpe portfolio concentrates almost entirely on three growth names: AMZN (43.5%), TSLA (29.7%), AAPL (16.4%), with T (4.4%) and GOOG (2.5%) holding only marginal positions. The outstanding stocks had a minor percentage in the portfolio allocation. Over the full period, this portfolio delivered a Sharpe ratio of 1.46 and turned $1 into roughly $22, versus about $2.50 for the S&P 500 over the same window, which is quite a sharp outperformance from passive index investing.

But we have to take the number with a grain of salt. The result from Task 6 was taken in-sample: the optimizer was handed the full 2012–2020 return history before choosing weights, effectively knowing in advance which stocks would perform best. This is an unrealistic setting for an active fund manager to be in. So, looking at Task 7, which provides a more real-world example by estimating our weights only on 2012–2015 data and applying them blindly to years 2016–2020. Then, the model results change. The expected an annualized return turns to be 37.7% and Sharpe ratio, which was based on Period 1 statistics, stands at 1.65. Yet, the realized performance on Period 2 had lower return, higher volatility, and a Sharpe ratio of only 1.12. This is a meaningful degradation from what the model has stated as expected return and the actual result. This strategy still beat the S&P 500 out of sample (roughly 3.5× vs 1.6×), but the gap between expected and realized performance is what needs to be kept in mind. The mean-variance optimization is only as good as the inputs it gets. The historical means, variances, and covariances are to be seen as estimates, not known constants. Small errors in those inputs can produce large, overconfident shifts in "optimal" weights.

And often portfolio rebalancing does not become the answer to solve this issue. The annual rebalancing experiment (re-optimizing every year using only the prior year's data) produced a highly inconsistent year-by-year Sharpe ratio (as low as ~-0.12 in weak years, as high as ~2.24 in strong ones), with a cumulative return that barely beat the S&P 500. We observe this trend as the stated one-year estimation window is short enough that the optimizer is effectively chasing whichever stock had a lucky prior year, and those winners frequently mean-revert or get disrupted by new conditions the model has no way of anticipating.

One caveat to note is that the task does not require to take into account issuer concentration limits, sector limits, or liquidity constraints a real regulated fund would have to respect. Due to this, most of the times the optimizer routes roughly 90% of the portfolio into three or four correlated tech/growth names. In real-world situations, portfolio managers do have guidelines and diversification quotas to meet, so this level of concentration is likely to be prohibited.

**Bottom line.** Applying the Markowitz mean-variance optimization to the following stocks does allow us to identify a portfolio that would have meaningfully outperformed a passive S&P 500 investment over 2012–2020. But the framework should be taken more as a directional signal where we see the optimizer preferring higher-growth, higher-volatility names like AMZN and TSLA rather than value/defensive names like T or IBM. The gap we observe in returns in Task 6's in-sample 1.46 and Task 7's out-of-sample 1.12 is the single most important number in this analysis: it's the honest price of not knowing the future outcomes and impact of unknown events.

## **Task 3:** Print out normalized (scaled) stock prices

**The idea:** divide every stock's price by its price on the first day (2012-01-12). So every stock and index included starts at 1.0 and you can more cleanly comare their growth trajectory.

```python
def normalize(df):
    # Keep Date out of the division, normalize only the price columns
    cols = [c for c in df.columns if c != "Date"]
    norm = df.copy()
    norm[cols] = df[cols] / df[cols].iloc[0]
    return norm

# Apply the normalization function to our dataframe
df_norm = normalize(df)

# Plot the normalized prices using the function defined in Task #2
def plot_normalized_prices(df_norm):
    cols_to_plot = [c for c in df_norm.columns if c != "Date"]
    plt.figure(figsize=(12, 6))
    
    for col in cols_to_plot:
        plt.plot(df_norm["Date"], df_norm[col], label=col, linewidth=0.8)
    
    # Reference line at 1.0 — anything above means positive return from start
    plt.axhline(y=1.0, color="black", linestyle="--", linewidth=0.8, label="Base (1.0)")
    
    plt.title("Normalized Stock Prices Change Over Time)")
    plt.xlabel("Date")
    plt.ylabel("Price relative to start (×)")
    plt.legend(loc="upper left", fontsize=8)
    plt.tight_layout()
    plt.show()

plot_normalized_prices(df_norm)
```

## **Task 4:** Calculate stock returns

```python
# Step 1
def calculate_daily_returns(data, stock_columns):
    """Computes the daily returns of every stock"""
    returns = pd.DataFrame()
    returns['Date'] = data['Date']

    for stock in stock_columns:
        stock_returns = []

        for i in range(len(data)):
            if i == 0:
                stock_returns.append(0)
            else:
                today_price = data[stock].iloc[i]
                yesterday_price = data[stock].iloc[i-1]
                daily_return = ((today_price - yesterday_price) / yesterday_price)*100
                stock_returns.append(daily_return)

        returns[stock] = stock_returns

    return returns
```

```python
# Step 2 — call the function and inspect:
df_returns = calculate_daily_returns(df, stock_cols + ['sp500'])

# Quick check: first row should be 0, second row should have values
print("First 3 rows of returns:")
print(df_returns.head(3).round(4))

# Summary statistics of daily returns
print("\nSummary statistics (daily returns):")
print(df_returns.drop(columns="Date").describe().round(4))
```

From the portfolio variance formula:

$$\sigma^2_p = \sum_i \sum_j w_i w_j \sigma_{ij}$$

where the covariance between two assets is $\sigma_{ij} = \rho_{ij} \sigma_i \sigma_j$.
A lower $\rho_{ij}$ directly shrinks the cross terms, reducing total portfolio risk.

**The correlation of MGM and BA:** We observe a correlation of 0.55 which is relatively high. The correlation is relatively high. This is a macro/cyclical co-movement (systematic risk), not an industry co-movement (idiosyncratic risk).

## **Task 5** : Calculate stock returns (Part 2)


```python

# - x-axis: the return value (e.g. -0.05 means -5% that day)
# - y-axis: how many days had a return in that range
# This tells us the shape of returns (are they symmetric or not, are there extreme days)? 

def plot_histograms(df_returns):
    # Get all columns except Date
    cols = [c for c in df_returns.columns if c != "Date"]
    
    # Set up a grid of subplots: 3 rows x 3 columns = 9 panels (one per asset)
    fig, axes = plt.subplots(3, 3, figsize=(14, 10))
    
    # Flatten the 2D array of axes into a 1D list for easy looping
    axes = axes.flatten()
    
    # Loop through each stock and plot its histogram
    for i, col in enumerate(cols):
        
        axes[i].hist(
            df_returns[col], 
            bins=80,                   # number of bars in the histogram
            color="steelblue",
            edgecolor="white",
            linewidth=0.3
        )
        
        # Add a vertical line at 0 to separate gains from losses
        axes[i].axvline(x=0, color="red", linestyle="--", linewidth=1)
        
        # Title and labels for each subplot
        axes[i].set_title(col, fontweight="bold")
        axes[i].set_xlabel("Daily Return")
        axes[i].set_ylabel("Frequency")
    
    plt.suptitle("Histograms of Daily Returns", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.show()


# Call the function
plot_histograms(df_returns)
```

## **Task 6** : Portfolio weights


```python
trading_days = 252

def simulate_portfolios(data, columns, trading_days, n_portfolios=10000, risk_free_rate=0.0): 

   results = []
   weights_record = []

   mean_returns = data[columns].mean() * trading_days   
   cov_matrix = data[columns].cov() * trading_days

   np.random.seed(7)
   for _ in range(n_portfolios):

       weights = np.random.dirichlet(np.ones(len(columns)))

       port_return = np.dot(weights, mean_returns)
       port_variance = weights @ cov_matrix @ weights
       port_volatility = np.sqrt(port_variance)
       sharpe = (port_return - risk_free_rate) / port_volatility

       results.append([port_return, port_variance, port_volatility, sharpe])
       weights_record.append(weights)

   results_df = pd.DataFrame(results, columns=['Return', 'Variance', 'Volatility', 'Sharpe'])

   max_sharpe_idx = results_df['Sharpe'].idxmax()
   max_sharpe_row = results_df.loc[max_sharpe_idx]
   optimal_weights = weights_record[max_sharpe_idx]

   return results_df, optimal_weights, max_sharpe_row

results_df, optimal_weights, max_sharpe_row = simulate_portfolios(data = df_returns, columns = stock_cols, trading_days =trading_days) #Task 6 answer
```

```python
print("=" * 52)
print("  TANGENCY PORTFOLIO (Maximum Sharpe Ratio)")
print("=" * 52)
print(f"  Annualized Return   : {max_sharpe_row['Return']:.2f}%")
print(f"  Annualized Variance : {max_sharpe_row['Variance']:.4f}")
print(f"  Annualized Std Dev  : {max_sharpe_row['Volatility']:.2f}%")
print(f"  Sharpe Ratio        : {max_sharpe_row['Sharpe']:.4f}")
print("\n  Weights:")

for stock, w in sorted(zip(stock_cols, optimal_weights), key=lambda x: x[1], reverse=True):
    print(f"    {stock:5s}: {w*100:.1f}%")
plt.figure(figsize=(12, 7))

scatter = plt.scatter(
    results_df['Volatility'],
    results_df['Return'],
    c=results_df['Sharpe'],
    cmap="viridis",
    alpha=0.4,
    s=8
)
plt.colorbar(scatter, label="Sharpe Ratio")

plt.scatter(
    max_sharpe_row['Volatility'],
    max_sharpe_row['Return'],
    color="red",
    marker="o",
    s=100,
    zorder=5,
    label=f"Tangency Portfolio  |  Sharpe = {max_sharpe_row['Sharpe']:.2f}"
)

plt.title("Empirical Efficient Frontier — 10,000 Simulated Portfolios", fontsize=14)
plt.xlabel("Annualized Volatility (Risk)")
plt.ylabel("Annualized Return")
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

**This graph** represents a scatter of 10,000 random portfolios in risk-return space, with purple representing those with a low sharpe ratio and those with yellow a high sharpe ratio. Here, the red dot represents the tangency portfolio, in other words the portfolio where the Sharpe ratio is maximised (return = 39.89%,Vol 27.41%, Sharpe 1.4557 with these corresponding weights for different stocks: AMZN 43.5%, TSLA 29.7%, AAPL 16.4%, T 4.4%, GOOG 2.5%, MGM 2.0%, IBM 1.0%, BA 0.5%).

**The 10,000 portfolios dont spread uniformly but form a bullet-like shape in the risk-return space**, representing the empirical frontier. The left/upper-left boudary of the cloud is considered the efficient frontier: for every level of risk, no simulated portfolio does better in return, therefore everything inside the cloud is dominated (you can get the same return with less risk, or more return for the same risk). This reflects one of Markowitz's result, that not all diversified portfolios are equally good, only those on the frontier are mean-variance efficient.

**The red dot:** the tangency portfolio is where we find the maximum Sharpe Ratio. This portfolio sits on the upper-left part of the frontier, not at the very top this is due to the fact that it trades off risk and return. Geometrically this is the point where the Capital Market Line (drawn from the origin since here rf=0) is tangeant to the frontier. This portfolio is the unique optimal risky portfolio for all mean-variance investors regardless of risk aversion, it can be combined with the riskfree asset to reach any efficient point.

```python
def plot_tangency_portfolio(
    daily_returns,
    tangency_weights,
    trading_days=252
):

    # Keep only the relevant return columns, indexed by date
    # df_returns is in percentage points (e.g. -1.93), so divide by 100 to get decimals for compounding
    returns = daily_returns.set_index("Date")[tangency_weights.index].copy() / 100

    # Portfolio daily returns, assuming daily rebalancing
    portfolio_daily_returns = returns.dot(tangency_weights)

    # Growth of $1 invested
    wealth_index = (1 + portfolio_daily_returns).cumprod()

    # Performance statistics
    cumulative_return = wealth_index.iloc[-1] - 1  # remove the initial dollar

    annualized_return = (
        wealth_index.iloc[-1]
        ** (trading_days / len(portfolio_daily_returns))
        - 1
    )

    annualized_volatility = (
        portfolio_daily_returns.std()
        * np.sqrt(trading_days)
    )

    # Sort weights for the bar chart
    display_weights = tangency_weights.sort_values()

    # Create both plots
    fig, axes = plt.subplots(
        1,
        2,
        figsize=(15, 6),
        gridspec_kw={"width_ratios": [1, 2]}
    )

    # Plot 1: Portfolio weights
    axes[0].barh(
        display_weights.index,
        display_weights.values,
        color="steelblue",
        edgecolor="white"
    )

    axes[0].set_title("Tangency Portfolio Weights")
    axes[0].set_xlabel("Portfolio Weight")
    axes[0].xaxis.set_major_formatter(
        mtick.PercentFormatter(xmax=1)
    )
    axes[0].grid(axis="x", alpha=0.25)

    # Add weight labels to the bars
    for i, weight in enumerate(display_weights.values):
        axes[0].text(
            weight + 0.005,
            i,
            f"{weight:.1%}",
            va="center"
        )

    # Plot 2: Historical performance
    axes[1].plot(
        wealth_index.index,
        wealth_index,
        color="navy",
        linewidth=1.6,
        label="Tangency Portfolio"
    )

    axes[1].axhline(
        y=1,
        color="gray",
        linestyle="--",
        linewidth=1,
        label="Initial Investment ($1)"
    )

    axes[1].set_title("Performance of $1 Invested")
    axes[1].set_xlabel("Date")
    axes[1].set_ylabel("Portfolio Value ($)")
    axes[1].legend()
    axes[1].grid(alpha=0.25)

    axes[1].xaxis.set_major_locator(
        mdates.YearLocator()
    )
    axes[1].xaxis.set_major_formatter(
        mdates.DateFormatter("%Y")
    )

    # Display performance in a box
    performance_text = (
        f"Cumulative return: {cumulative_return:.2%}\n"
        f"Annualized return: {annualized_return:.2%}\n"
        f"Annualized volatility: {annualized_volatility:.2%}"
    )

    axes[1].text(
        0.02,
        0.97,
        performance_text,
        transform=axes[1].transAxes,
        va="top",
        bbox=dict(
            boxstyle="round",
            facecolor="white",
            edgecolor="lightgray",
            alpha=0.9
        )
    )

    fig.tight_layout()
    plt.show()

    performance_summary = pd.Series({
        "Cumulative Return": cumulative_return,
        "Annualized Return": annualized_return,
        "Annualized Volatility": annualized_volatility
    })

    return performance_summary, portfolio_daily_returns, wealth_index


tangency_weights = pd.Series(optimal_weights, index=stock_cols)

performance, portfolio_returns, wealth = plot_tangency_portfolio(df_returns, tangency_weights, trading_days)

print("Tangency portfolio weights:")
print(tangency_weights.apply(lambda x: f"{x:.2%}"))

print("\nPortfolio performance:")
print(performance.apply(lambda x: f"{x:.2%}"))
```

_______

**The optimizer:** The optimizer loads up heavily on the three best performing stocks over the given period (AMZN, TSLA, AAPL). Nearly 90% of the portfolio is allocated to the top 3 stocks, while the other stocks represent only 10% (the optimizer essentially dismisses them). ### ***Optional:*** Define a function that finds the weights that minimize the variance for all possible level or return 

```python
mean_returns = df_returns[stock_cols].mean() * trading_days  # annualize daily mean
cov_matrix = df_returns[stock_cols].cov() * trading_days

def portfolio_variance(weights, cov_matrix):
    return weights @ cov_matrix @ weights

def portfolio_return(weights, mean_returns):
    return np.dot(weights, mean_returns)

def efficient_frontier(mean_returns, cov_matrix, n_points=50):
    n_assets = len(mean_returns)
    target_returns = np.linspace(mean_returns.min(), mean_returns.max(), n_points)

    frontier_volatility = []
    frontier_weights = []

    bounds = tuple((0, 1) for _ in range(n_assets))
    init_guess = np.ones(n_assets) / n_assets

    for target in target_returns:
        constraints = (
            {'type': 'eq', 'fun': lambda w: np.sum(w) - 1},
            {'type': 'eq', 'fun': lambda w, target=target: portfolio_return(w, mean_returns) - target}
        )

        result = minimize(portfolio_variance, init_guess, args=(cov_matrix,),
                           method='SLSQP', bounds=bounds, constraints=constraints)

        frontier_volatility.append(np.sqrt(result['fun']))
        frontier_weights.append(result['x'])

    return target_returns, frontier_volatility, frontier_weights

target_returns, frontier_vol, frontier_w = efficient_frontier(mean_returns, cov_matrix)

plt.figure(figsize=(10, 6))
plt.scatter(results_df['Volatility'], results_df['Return'], c=results_df['Sharpe'], cmap="viridis", alpha=0.4)
plt.colorbar(label='Sharpe Ratio')
plt.plot(frontier_vol, target_returns, color='purple', linewidth=2)
plt.scatter(max_sharpe_row['Volatility'], max_sharpe_row['Return'], color='red', s=100)
plt.xlabel('Volatility')
plt.ylabel('Return')
plt.title('Efficient Frontier')
plt.show()
```

**Construction and Interpretation of the Efficient Frontier**

To construct the portfolio set, we first estimated each asset’s expected annual return and the annual covariance matrix from daily historical returns. Expected returns were annualized by multiplying the average daily returns by 252, while the covariance matrix was also multiplied by 252. For any vector of portfolio weights, the expected return and variance are then given by:

$$
E(R_p)=\mathbf{w}^{T}\boldsymbol{\mu}
$$

$$
\sigma_p^2=\mathbf{w}^{T}\boldsymbol{\Sigma}\mathbf{w}
$$

- The code selects 50 target returns between the lowest and highest individual expected asset returns. 
- For each target, the SLSQP optimizer finds the portfolio with the lowest possible variance, subject to three conditions: 
    - the weights must sum to one, 
    - the portfolio must achieve the specified target return, and 
    - every weight must remain between zero and one to prevent short selling. 
Taking the square root of the minimized variance gives the portfolio volatility plotted on the horizontal axis.

The section above the global minimum-variance portfolio is the efficient frontier, the purple line in the graph above. Portfolios on this upper section offer the highest expected return for a given level of risk, meaning that no other feasible portfolio can provide a higher expected return without also increasing volatility. The lower section is part of the minimum-variance boundary but is inefficient because it can be dominated by portfolios offering higher returns for similar risk.


**Tangency Portfolio**

The black point represents the portfolio with the highest Sharpe ratio found among the 10,000 simulated portfolios. It approximates the tangency portfolio, which offers the greatest expected excess return per unit of volatility:

$$
\text{Sharpe ratio}
=
\frac{E(R_p)-R_f}{\sigma_p}
$$

Because the black point was selected from a finite random sample, it may lie slightly inside the optimized frontier rather than exactly on it. Increasing the number of simulations would generally bring the best simulated portfolio closer to the theoretical tangency point, but this can be computationally inefficient and does not guarantee an exact result. A more precise approach would maximize the Sharpe ratio directly using constrained numerical optimization.

**Key Takeaway**

The two approaches are complementary. Random simulation gives an intuitive picture of the feasible portfolio set and demonstrates how different allocations produce different combinations of risk and return. Constrained optimization provides a more precise estimate of the minimum-variance boundary and identifies the portfolios that make the most efficient use of risk.

The graph supports the central idea of Markowitz theory: diversification is not simply about holding many assets, but about combining assets according to their expected returns, volatilities, and correlations. The most attractive portfolios lie on the upper part of the optimized boundary, while portfolios inside the cloud are inefficient because another feasible allocation can provide a higher expected return for the same risk or lower risk for the same expected return.

**Interpretation:**

True Efficient Frontier vs. Simulated Portfolios

The black curve represents the true efficient frontier — computed by mathematically minimizing variance for every possible return level using constrained optimization. It forms the theoretical left boundary of the feasible set: no portfolio can ever sit to the left of this line, and the plot confirms this: all 10,000 random portfolios fall to its right.

The gap between the black curve and the cloud of dots  illustrates an important limitation of the simulation approach: random sampling is inefficient at finding the extreme edges of the weight space. This is because concentrated portfolios (e.g. 90% in one stock) are extremely rare among random draws, yet these are often the ones that define the frontier. The optimizer finds them exactly, while random simulation almost never lands there.

The tangency portfolio (red dot) sits slightly inside the frontier for the same reason: it is the best portfolio found among 10,000 random draws, but not the mathematically exact tangency point. With millions of simulations it would converge closer to the curve.

The key takeaway is that simulation gives a good approximation of the feasible set, but mathematical optimization is necessary to trace the true efficient frontier precisely. Both approaches agree on the overall shape and confirm that the best risk-return tradeoffs lie along the upper-left boundary of the cloud.

### ***Optional:*** Part 7 - Annual Rebalancing Strategy

```python
# OPTIONAL — Annual Rebalancing Strategy
# Instead of fixing weights once in 2012 and holding forever, we re-optimize every year using only the previous year's data:
#   - estimate the tangency weights on year Y
#   - apply those weights during year Y+1 and collect the daily returns

all_years = sorted(df_returns["Date"].dt.year.unique())
estimation_years  = all_years[:-1]     # years used to estimate the weights
application_years = all_years[1:]      # years where those weights are applied

rebalanced_daily = []     # daily returns of the strategy, year after year
yearly_summary = []

for est_year, app_year in zip(estimation_years, application_years):

    ret_est = df_returns[df_returns["Date"].dt.year == est_year]
    ret_app = df_returns[df_returns["Date"].dt.year == app_year]

    # Skip incomplete years
    if len(ret_est) < 50:
        continue

    # Estimation: tangency weights based on last year only
    _, weights_year, best_row = simulate_portfolios(ret_est, stock_cols, trading_days)

    # Application: those same weights, on the following year
    port_daily = ret_app[stock_cols] @ weights_year
    rebalanced_daily.append(port_daily)

    yearly_summary.append({
        "est_year": est_year,
        "app_year": app_year,
        "sharpe_expected": best_row['Sharpe'],
        "ann_return": port_daily.mean() * trading_days,
        "ann_vol": port_daily.std() * np.sqrt(trading_days),
        "top_holding": pd.Series(weights_year, index=stock_cols).idxmax(),
        "top_weight": pd.Series(weights_year, index=stock_cols).max()
    })

summary_df = pd.DataFrame(yearly_summary)
summary_df['sharpe_realized'] = summary_df['ann_return'] / summary_df['ann_vol']

print("=== ANNUAL REBALANCING — YEAR BY YEAR ===")
print(summary_df.round(3).to_string(index=False))
```

```python
# Overall performance: rebalanced strategy vs a naive 1/N portfolio vs the market

strategy = pd.concat(rebalanced_daily)
dates = df_returns.loc[strategy.index, "Date"]

# Equally weighted portfolio: 1/8 in each stock, no optimization at all
equal_weights = np.ones(len(stock_cols)) / len(stock_cols)
equal = df_returns.loc[strategy.index, stock_cols] @ equal_weights

market = df_returns.loc[strategy.index, "sp500"]

plt.figure(figsize=(12, 6))
plt.plot(dates, (1 + strategy / 100).cumprod(), label="Annual rebalancing (Markowitz)", linewidth=1.5)
plt.plot(dates, (1 + equal / 100).cumprod(), label="Equally weighted (1/N)", linewidth=1.5)
plt.plot(dates, (1 + market / 100).cumprod(), label="S&P 500", linewidth=1.5, linestyle="--")

# Mark each rebalancing date
for year in summary_df["app_year"]:
    plt.axvline(pd.Timestamp(f"{year}-01-01"), color="grey", linestyle=":", linewidth=0.8, alpha=0.6)

plt.axhline(y=1, color="black", linestyle=":", linewidth=0.6)
plt.title("Annual Rebalancing vs a Naive 1/N Portfolio vs the Market")
plt.xlabel("Date")
plt.ylabel("Growth of $1 invested")
plt.legend()
plt.tight_layout()
plt.show()

def performance(returns, name):
    ret = returns.mean() * trading_days
    vol = returns.std() * np.sqrt(trading_days)
    return {'Strategy': name, 'Ann. return (%)': ret, 'Ann. volatility (%)': vol,
            'Sharpe': ret / vol, 'Growth of $1': (1 + returns / 100).cumprod().iloc[-1]}

comparison = pd.DataFrame([
    performance(strategy, 'Annual rebalancing (Markowitz)'),
    performance(equal, 'Equally weighted (1/N)'),
    performance(market, 'S&P 500')
]).set_index('Strategy')

print(comparison.round(3).to_string())
```

Its Sharpe ratio is therefore lower (0.59 vs 0.74), meaning the small return advantage is not worth the extra risk taken.
